```python

# import pandas as pd
# from difflib import SequenceMatcher

# def similarity_score(str1: str, str2: str) -> float:
#     return SequenceMatcher(None, str1, str2).ratio() * 100

# def align_dataframe_columns(df_primary: pd.DataFrame, df_recipient: pd.DataFrame, threshold: float = 80.0):
#     df_recipient_aligned = df_recipient.copy()
    
#     primary_cols = list(df_primary.columns)
#     recipient_cols = list(df_recipient_aligned.columns)
    
#     column_mapping = {}
#     matched_recipient_cols = set()
#     matching_records = []

#     for p_col in primary_cols:
#         best_match = None
#         best_score = 0.0
        
#         available_recipient_cols = [c for c in recipient_cols if c not in matched_recipient_cols]
        
#         for r_col in available_recipient_cols:
#             score = similarity_score(p_col, r_col)
#             if score > best_score:
#                 best_score = score
#                 best_match = r_col
                
#         if best_match and best_score >= threshold:
#             column_mapping[best_match] = p_col
#             matched_recipient_cols.add(best_match)
#             matching_records.append({
#                 'ColumnNamePrimary': p_col,
#                 'ColumnNameRecipient': best_match,
#                 'SimilarityScore': round(best_score, 2)
#             })

#     # Dataframe of successful matches
#     df_matching = pd.DataFrame(matching_records, columns=['ColumnNamePrimary', 'ColumnNameRecipient', 'SimilarityScore'])   
    
#     # Identify unmatched columns from both primary and recipient dataframes
#     unmatched_recipient_cols = [c for c in recipient_cols if c not in matched_recipient_cols]
#     missing_primary_cols = [c for c in primary_cols if c not in column_mapping.values()]
    
#     # Build df_unmatching dataframe
#     df_unmatching = pd.DataFrame({
#         'ColumnNamePrimary': pd.Series(missing_primary_cols, dtype='object'),
#         'ColumnNameRecipient': pd.Series(unmatched_recipient_cols, dtype='object')
#     })
    
#     # Dataframe of dropped recipient columns
#     df_drop = df_recipient_aligned[unmatched_recipient_cols].copy()
    
#     # Rename matched columns
#     df_recipient_aligned.rename(columns=column_mapping, inplace=True)
    
#     # Remove duplicate column names if any exist after renaming
#     df_recipient_aligned = df_recipient_aligned.loc[:, ~df_recipient_aligned.columns.duplicated()].copy()
    
#     # Add missing primary columns filled with None
#     for m_col in missing_primary_cols:
#         df_recipient_aligned[m_col] = None
        
#     df_add = df_recipient_aligned[missing_primary_cols].copy()
    
#     # Safely reorder columns to match df_primary layout
#     df_recipient_aligned = df_recipient_aligned.reindex(columns=primary_cols)
    
#     return df_recipient_aligned, df_drop, df_add, df_matching, df_unmatching

import pandas as pd
from difflib import SequenceMatcher

def similarity_score(str1: str, str2: str) -> float:
    return SequenceMatcher(None, str1, str2).ratio() * 100

def align_dataframe_columns(df_primary: pd.DataFrame, df_recipient: pd.DataFrame, threshold: float = 80.0):
    df_recipient_aligned = df_recipient.copy()
    
    primary_cols = list(df_primary.columns)
    recipient_cols = list(df_recipient_aligned.columns)
    
    # 1. Evaluate all combinations of primary and recipient columns globally
    all_pairs = []
    for p_col in primary_cols:
        for r_col in recipient_cols:
            score = similarity_score(p_col, r_col)
            all_pairs.append({
                'ColumnNamePrimary': p_col,
                'ColumnNameRecipient': r_col,
                'SimilarityScore': score
            })
    
    # Sort pairs by score descending to prioritize the best global matches first
    all_pairs.sort(key=lambda x: x['SimilarityScore'], reverse=True)
    
    column_mapping = {}
    matched_primary = set()
    matched_recipient = set()
    matching_records = []
    
    for pair in all_pairs:
        p_col = pair['ColumnNamePrimary']
        r_col = pair['ColumnNameRecipient']
        score = pair['SimilarityScore']
        
        if score < threshold:
            break  # Since sorted descending, all subsequent scores are below threshold
            
        if p_col not in matched_primary and r_col not in matched_recipient:
            matched_primary.add(p_col)
            matched_recipient.add(r_col)
            column_mapping[r_col] = p_col  # Map recipient column name to primary column name
            matching_records.append({
                'ColumnNamePrimary': p_col,
                'ColumnNameRecipient': r_col,
                'SimilarityScore': round(score, 2)
            })

    # 2. Build df_matching and sort from lowest to highest score correctly
    df_matching = pd.DataFrame(matching_records, columns=['ColumnNamePrimary', 'ColumnNameRecipient', 'SimilarityScore'])
    if not df_matching.empty:
        df_matching = df_matching.sort_values(by='SimilarityScore', ascending=True).reset_index(drop=True)
    
    # Identify unmatched columns from both primary and recipient dataframes
    unmatched_recipient_cols = [c for c in recipient_cols if c not in matched_recipient]
    missing_primary_cols = [c for c in primary_cols if c not in matched_primary]
    
    # Build df_unmatching dataframe
    df_unmatching = pd.DataFrame({
        'ColumnNamePrimary': pd.Series(missing_primary_cols, dtype='object'),
        'ColumnNameRecipient': pd.Series(unmatched_recipient_cols, dtype='object')
    })
    
    # Dataframe of dropped recipient columns
    df_drop = df_recipient_aligned[unmatched_recipient_cols].copy()
    
    # Rename matched columns
    df_recipient_aligned.rename(columns=column_mapping, inplace=True)
    
    # Remove duplicate column names if any exist after renaming
    df_recipient_aligned = df_recipient_aligned.loc[:, ~df_recipient_aligned.columns.duplicated()].copy()
    
    # Add missing primary columns filled with None
    for m_col in missing_primary_cols:
        df_recipient_aligned[m_col] = None
        
    df_add = df_recipient_aligned[missing_primary_cols].copy()
    
    # Safely reorder columns to match df_primary layout
    df_recipient_aligned = df_recipient_aligned.reindex(columns=primary_cols)
    
    return df_recipient_aligned, df_drop, df_add, df_matching, df_unmatching
```

```python
import pandas as pd

def clean_and_assign_dtypes(df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Cleans and standardizes DataFrame data types into universal types (str, int, float, date).
    Compatible with Pandas, Excel, and SQL databases like Supabase.
    
    Returns:
        df: Cleaned DataFrame with standard data types.
        df_error: DataFrame containing unique unparseable values labelled by 
                  'ColumnName (target_type)'.
    """
    df = df.copy()
    errors_dict = {}

    # 1. DATE COLUMNS (Target: date / datetime)
    date_cols = [
        'Date', 'RegistrarDate', 'Tx_Outcome_Date', 
        'DOTStartedDate111', 'TPTregistrationdate', 'Drugpickup1', 
        'Date_drug_pickup2', 'Date_drug_pickup_3', 'Date_drug_pickup_4', 
        'Date_drug_pickup_5', 'Date_drug_pickup_6', 'Date_drug_pickup_7', 
        'Date_drug_pickup_8', 'Date_drug_pickup_9', 'TPTdiscontinuationdate', 
        'OutcomeDate','CPT', 'ART'
    ]

    for col in date_cols:
        if col in df.columns:
            parsed = pd.to_datetime(df[col], errors='coerce')
            
            # Identify values that failed date conversion
            invalid_mask = df[col].notna() & parsed.isna()
            invalid_values = df.loc[invalid_mask, col].unique().tolist()
            if invalid_values:
                errors_dict[f"{col} (date)"] = invalid_values
            
            df[col] = parsed

    # 2. STRING COLUMNS (Target: str)
    # Includes all previously categorical fields + standard text columns
    string_cols = [
        # Categorical text fields converted to str
        'StateRegionName', 'Tsp', 'WardVillage', 'WardCode', 'Approach', 
        'Ptstsp', 'Nationality', 'Organization', 
        'Transferin', 'TypeofPatients', 'TypeofDisease', 'HIVStatus', 
        'XDRResult', 'OLDTYPEOFTBTREATMENT', 'Modifiedregimen', 
        'CultureDSTresult', 'Microscope_Result_2nd_mnth', 'Xpert_Result_2nd_mnth', 
        'Microscope_Result_3rd_mnth', 'Xpert_Result_3rd_mnth', 
        'Microscope_Result_5th_mnth', 'Xpert_Result_5th_mnth', 
        'EndofTxMicroscopeResult', 'EndofTxXRayResult', 'EndofTxXpertResult', 
        'TreatmentOutcome1211', 'VolunteerGender111', 'VolunteerOrganization111', 
        'ContactInvestigation111', 'DOTSupervision111', 
        'DOTsupervisiontillTreatmentComp111', 'MonthDiagnosis11', 'Location11', 
        'EPI11', 'TB11', 'XrayRemark11', 'ReportingTsp11', 'XrayEQA11', 
        'TPTregimen', 'TPTTreatmentOutcome',
        # Regular text / ID fields
        'ReportingPeriod', 'TownshipTBRegNumber', 'PatientID', 'Name', 'Address', 
        'PhoneNo', 'VolunteerName', 'SpecifyifEPSiteofEPTB', 'Other1', 'Cxr_others', 
        'Ref1', 'DOTvolName111', 'Remark', 'TPTregno', 'RegnoofindexTBcase', 
        'DrugDose'
    ]

    for col in string_cols:
        if col in df.columns:
            # Cast safely to standard python string, retaining empty/NaN entries correctly
            df[col] = df[col].astype("string")

    # 3. INTEGER COLUMNS (Target: int)
    int_cols = [
        'Team', 'Visit_no', 'Sr_No', 'Age', 'Sex', 'Cough', 'Fever', 'Wtloss', 
        'Nightsweat', 'Haemoptysis', 'Chestpain', 'Fatigue', 'Neckglands', 
        'TBcontact', 'MDRTBcontact', 'TBTreatmenthistory', 'Smoking', 
        'VOL', 'PublicHealthCare1', 'ReportingYear11', 'UniqueKey',
        'Referralfor', 'Reasonforexamination', 'TypeofPatient', 'TypeofPatient1', 
        'DM1', 'HT1', 'DMHT1', 'RTIAVI1', 'Generalweakness1', 'Cxrr', 
        'CXRresult', 'Sputum_request', 'Micror', 'Sputummicroscopyresult', 
        'Genexpertrequested', 'GeneXpertresult', 'Bact_status', 'Case', 
        'Treatmentreferral', 'TreatmentRegimen', 'Placeforreferral', 
        'Seeing1', 'Hearing1', 'Walking1', 'Cognition1', 'Selfcare1', 
        'Communication1', 'Disability1', 'Xray2ndReading11', 'CXRresult211', 
        'Recstatus', 'TypeofTBTreatment','Origin'
    ]

    for col in int_cols:
        if col in df.columns:
            parsed = pd.to_numeric(df[col], errors='coerce')
            invalid_mask = df[col].notna() & parsed.isna()
            invalid_values = df.loc[invalid_mask, col].unique().tolist()
            if invalid_values:
                errors_dict[f"{col} (int)"] = invalid_values

            # Uses 'Int64' to support integer values alongside missing/null entries
            df[col] = parsed.astype('Int64')

    # 4. FLOAT COLUMNS (Target: float)
    float_cols = ['RBS', 'HbA1c', 'Systolic', 'Diastolic']

    for col in float_cols:
        if col in df.columns:
            parsed = pd.to_numeric(df[col], errors='coerce')
            invalid_mask = df[col].notna() & parsed.isna()
            invalid_values = df.loc[invalid_mask, col].unique().tolist()
            if invalid_values:
                errors_dict[f"{col} (float)"] = invalid_values

            df[col] = parsed.astype('float64')

    # Build df_error showing invalid unique values per field type
    if errors_dict:
        df_error = pd.DataFrame(dict([(k, pd.Series(v)) for k, v in errors_dict.items()]))
    else:
        df_error = pd.DataFrame()

    return df, df_error
```

```python
DATE_COLUMNS = ["Date","RegistrarDate","CPT","ART","Tx_Outcome_Date","DOTStartedDate111","TPTregistrationdate",
    "Drugpickup1","Date_drug_pickup2","Date_drug_pickup_3","Date_drug_pickup_4","Date_drug_pickup_5","Date_drug_pickup_6",
    "Date_drug_pickup_7","Date_drug_pickup_8","Date_drug_pickup_9","TPTdiscontinuationdate","OutcomeDate"]

def functionGetDataFromTable(tableName: str, url: str, key: str, page_size: int = 1000) -> list:
    endpoint = f"{url.rstrip('/')}/rest/v1/{tableName}?select=*"
    headers = {
        "apikey": key.strip(),
        "Authorization": f"Bearer {key.strip()}",
        "Content-Type": "application/json",
    }
    all_data = []
    start_index = 0
    try:
        while True:
            page_headers = headers.copy()
            page_headers["Range"] = f"{start_index}-{start_index + page_size - 1}"
            response = requests.get(endpoint, headers=page_headers)
            response.raise_for_status()
            chunk = response.json()
            if not chunk:
                break  # Stop when no more rows are returned
            all_data.extend(chunk)
            if len(chunk) < page_size:
                break
            start_index += page_size
        print(f"✅ Successfully retrieved ALL {len(all_data)} rows from '{tableName}'")
        return pd.DataFrame(all_data)
    except requests.exceptions.HTTPError as http_err:
        print(f"❌ HTTP error occurred: {http_err}\nResponse details: {response.text}")
    except Exception as err:
        print(f"❌ An error occurred: {err}")
    return None

def functionUpdateDataFromDataframe(tableName: str,Dataframe: pd.DataFrame,url: str,key: str,date_columns: list = None,primary_key: str = "PatientID",batch_size: int = 1000,) -> int:

    date_columns = DATE_COLUMNS
    if date_columns is None:
        date_columns = []
    if Dataframe.empty:
        print("⚠️ Provided DataFrame is empty. No records to update.")
        return 0
    df_update = Dataframe.copy()
    for col in date_columns:
        if col in df_update.columns:
            df_update[col] = pd.to_datetime(df_update[col], errors="coerce")
            df_update[col] = df_update[col].dt.strftime("%Y-%m-%d")  # Ensures plain Python string or NaN
    for col in df_update.columns:
        if col not in date_columns:
            if df_update[col].dtype == "float64":
                try:
                    df_update[col] = df_update[col].astype("Int64")
                except Exception:
                    pass
    records = df_update.to_dict(orient="records")
    clean_records = []
    for record in records:
        clean_record = {}
        for k, v in record.items():
            if pd.isna(v):
                clean_record[k] = None
            elif isinstance(v, (pd.Timestamp, pd.Timedelta)):
                clean_record[k] = v.strftime("%Y-%m-%d")
            elif isinstance(v, float) and v.is_integer():
                clean_record[k] = int(v)
            else:
                clean_record[k] = v
        clean_records.append(clean_record)
    total_records = len(clean_records)
    endpoint = f"{url.rstrip('/')}/rest/v1/{tableName}?on_conflict={primary_key}"
    headers = {
        "apikey": key.strip(),
        "Authorization": f"Bearer {key.strip()}",
        "Content-Type": "application/json",
        "Prefer": "resolution=merge-duplicates,return=minimal",
    }
    total_updated = 0
    num_batches = math.ceil(total_records / batch_size)
    print(f"📦 Upserting {total_records} records in {num_batches} batch(es)...")
    for i in range(0, total_records, batch_size):
        chunk = clean_records[i : i + batch_size]
        batch_num = (i // batch_size) + 1
        try:
            payload = json.dumps(chunk)
            response = requests.post(endpoint, headers=headers, data=payload)
            response.raise_for_status()
            total_updated += len(chunk)
            print(f"  ✓ Batch {batch_num}/{num_batches} ({len(chunk)} rows) successful.")
        except requests.exceptions.HTTPError as http_err:
            print(f"❌ HTTP error on batch {batch_num}: {http_err}\nResponse: {response.text}")
            return total_updated
        except Exception as err:
            print(f"❌ Error on batch {batch_num}: {err}")
            return total_updated

    print(f"✅ Successfully upserted {total_updated} / {total_records} record(s) into '{tableName}'")
    return total_updated

def functionDeleteRowByPatientID(tableName: str, patient_id_list: list, url: str, key: str) -> dict:
    if not patient_id_list:
        print("⚠️ PatientID list is empty. Nothing to delete.")
        return None
    formatted_ids = ",".join(map(str, patient_id_list))
    endpoint = f"{url.rstrip('/')}/rest/v1/{tableName}?PatientID=in.({formatted_ids})"
    headers = {
        "apikey": key.strip(),
        "Authorization": f"Bearer {key.strip()}",
        "Content-Type": "application/json",
        # 'return=representation' instructs Supabase to return the deleted rows
        "Prefer": "return=representation",
    }
    try:
        response = requests.delete(endpoint, headers=headers)
        response.raise_for_status()

        deleted_data = response.json()
        print(f"✅ Successfully deleted {len(deleted_data)} row(s) from '{tableName}'")
        return pd.DataFrame(deleted_data)
    except requests.exceptions.HTTPError as http_err:
        print(f"❌ HTTP error occurred: {http_err}\nResponse details: {response.text}")
    except Exception as err:
        print(f"❌ An error occurred: {err}")
    return None

def check_duplicate_patient_ids(df: pd.DataFrame, id_col: str = "PatientID") -> pd.DataFrame:
    duplicate_mask = df.duplicated(subset=[id_col], keep=False)
    duplicates_df = df[duplicate_mask]
    num_duplicates = df[id_col].duplicated().sum()
    print(f"🔍 Found {num_duplicates} duplicate PatientID occurrence(s).")
    return duplicates_df

def functionUnmatchingID(df1: pd.DataFrame, df2: pd.DataFrame, id_col: str = "PatientID") -> pd.DataFrame:
  frame = inspect.currentframe().f_back
  frame_info = inspect.getframeinfo(frame)
  df1_name, df2_name = "df1", "df2"
  if frame_info.code_context:
    call_line = frame_info.code_context[0].strip()
    try:
      # Parse the Python expression into an AST node
      parsed = ast.parse(call_line).body[0]
      if isinstance(parsed, ast.Expr) and isinstance(parsed.value, ast.Call):
        args = parsed.value.args
        if len(args) >= 1 and isinstance(args[0], ast.Name):
          df1_name = args[0].id
        if len(args) >= 2 and isinstance(args[1], ast.Name):
          df2_name = args[1].id
    except Exception:
      pass  # Fallback to default names if parsing fails (e.g. multi-line calls)
  col1_name = f"ID_{df1_name}"
  col2_name = f"ID_{df2_name}"
  s1_ids = df1[id_col].dropna().unique()
  s2_ids = df2[id_col].dropna().unique()
  s1 = pd.DataFrame({id_col: s1_ids, col1_name: s1_ids})
  s2 = pd.DataFrame({id_col: s2_ids, col2_name: s2_ids})
  merged = pd.merge(s1, s2, on=id_col, how="outer").fillna("")
  merged = merged.rename(columns={id_col: "ID"})
  unmatched_df = merged[(merged[col1_name] == "") | (merged[col2_name] == "")].reset_index(drop=True)

  print(f"⚠️ Found {len(unmatched_df)} unmatched ID(s) between '{df1_name}' and"
        f" '{df2_name}'.")
  return unmatched_df[["ID", col1_name, col2_name]]

def normalize_value(val):
    if pd.isna(val) or val is None or val is np.nan:
        return None
    s_val = str(val).strip()
    if s_val == "" or s_val.lower() in ("nan", "nat", "none", "null"):
        return None
    try:
        f_val = float(s_val)
        if f_val.is_integer():
            return str(int(f_val))
        return str(f_val)
    except (ValueError, TypeError):
        pass
    return s_val

def compare_data(dfExcel: pd.DataFrame, dfSupabase: pd.DataFrame, id_col: str = "PatientID",date_cols: list = DATE_COLUMNS) -> pd.DataFrame:
    records = []
    dfE = dfExcel.copy()
    dfS = dfSupabase.copy()
    dfE[id_col] = dfE[id_col].astype(str).str.strip()
    dfS[id_col] = dfS[id_col].astype(str).str.strip()
    for col in date_cols:
        if col in dfE.columns:
            dfE[col] = pd.to_datetime(dfE[col], errors="coerce").dt.strftime("%Y-%m-%d")
        if col in dfS.columns:
            dfS[col] = pd.to_datetime(dfS[col], errors="coerce").dt.strftime("%Y-%m-%d")
    ids_dfExcel = set(dfE[id_col].dropna().unique())
    ids_dfSupabase = set(dfS[id_col].dropna().unique())
    for pid in ids_dfExcel - ids_dfSupabase:
        records.append([pid, "_Not in Supabase", pid, None])
    for pid in ids_dfSupabase - ids_dfExcel:
        records.append([pid, "_Not in Excel", None, pid])
    cols_dfExcel = set(dfE.columns) - {id_col}
    cols_dfSupabase = set(dfS.columns) - {id_col}
    for col in cols_dfExcel - cols_dfSupabase:
        records.append([col, "Missing Column in dfSupabase", "Exists", "Missing"])
    for col in cols_dfSupabase - cols_dfExcel:
        records.append([col, "Missing Column in dfExcel", "Missing", "Exists"])
    common_cols = list((set(dfE.columns) & set(dfS.columns)) - {id_col})

    if common_cols:
        merged = pd.merge(
            dfE[[id_col] + common_cols],
            dfS[[id_col] + common_cols],
            on=id_col,
            how="inner",
            suffixes=("_Excel", "_Supabase"),
        )
        for col in common_cols:
            col_e = f"{col}_Excel"
            col_s = f"{col}_Supabase"
            for _, row in merged.iterrows():
                pid = row[id_col]
                norm1 = normalize_value(row[col_e])
                norm2 = normalize_value(row[col_s])
                if norm1 is None and norm2 is None:
                    continue
                if norm1 != norm2:
                    records.append([pid, col, row[col_e], row[col_s]])
    return pd.DataFrame(records, columns=["PatientID", "ColumnName", "Value_Excel", "Value_Supabase"])
```
